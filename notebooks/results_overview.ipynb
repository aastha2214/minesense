# MineSense: results walkthrough

This notebook only *reads* the tables and figures written by the pipeline (`python -m minesense.<step>`, see the README). It trains nothing, so it runs in seconds.

```python
import json
from pathlib import Path

import pandas as pd
from IPython.display import Image, display

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
T, F = ROOT / 'results' / 'tables', ROOT / 'results' / 'figures'
```

## Data audit and the height law

```python
audit = json.loads((T / 'data_audit.json').read_text())
print(audit['quality']['class_counts'])
print(audit['design'])
display(Image(F / 'fig05_voltage_vs_height_by_class.png', width=700))
```

## Nested cross-validation (25 test folds)

```python
summary = pd.read_csv(T / 'stage3_summary.csv')
summary.sort_values('accuracy', ascending=False)[['model', 'accuracy', 'accuracy_std', 'macro_f1', 'detection', 'mine_missed', 'false_alarm']].round(3)
```

```python
sig = pd.read_csv(T / 'significance_tests.csv')
sig[sig.family == 'best neural vs baseline'][['model_b', 'metric', 'mean_diff', 'wins', 'losses', 'p_wilcoxon_holm', 'p_corrected_t_holm']]
```
